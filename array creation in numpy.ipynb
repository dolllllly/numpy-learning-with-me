```python
arr.shape   #rows and column
```

```python
arr.size    #total element
```

```python
arr.dtype     #datatype
```

```python
Array methods
```

```python
arr.min()    #minimum element
```

```python
arr.max()       #maximum element
```

```python
arr.sum()     #sum of all element
```

```python
np.sum(arr,axis=0)         #column wise summision then axis=0
```

```python
np.sum(arr,axis=1)     #row wise summision the axis=1
```

```python
arr.mean()           #mean element
```

```python
arr.std()        # standard diviation
```

```python
arr.argmax()        # index of the max element
```

```python
arr.argmin()          #index of the min element
```
